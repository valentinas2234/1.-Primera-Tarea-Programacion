```python
# PROGRAMA: CALCULADORA DE RENDIMIENTO AGROINDUSTRIAL
# Autor: Valentina Salgado Castillo - Ingeniería Agroindustrial
# Objetivo: Calcular litros de jugo basado en rendimiento

# Módulo 1: Interfaz de usuario
print("""

SISTEMA DE CÁLCULO DE RENDIMIENTO EN JUGOS DE FRUTAS Versión 1.0 (2024)""")

# Módulo 2: Captura de datos
fruta_kg = input("\nIngrese la cantidad de fruta procesada (kg): ")
rendimiento_porcentaje = input("Ingrese el % de rendimiento esperado (Ej: 55.5): ")

# Módulo 3: Procesamiento numérico
try:
    kg = float(fruta_kg)  # Conversión a número decimal
    rend = float(rendimiento_porcentaje)
    jugo_litros = (kg * rend) / 100  # Fórmula central
except ValueError:
    print("Error: Solo se permiten valores numéricos")
    exit()

# Módulo 4: Presentación de resultados
print("\n" + "="*40)
print(f"ANÁLISIS DE RENDIMIENTO")
print("="*40)
print(f"- Materia prima: {kg} kg de fruta")
print(f"- Eficiencia técnica: {rend}%")
print("-"*40)
print(f"JUGUO PRODUCIDO: {jugo_litros:.2f} litros")
print("="*40)
```
